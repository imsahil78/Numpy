BROADCASTING 

allows  numpy to perfrom op on array with diff shapes by virtually expanding shape
Only work if number is same or one should be 1

```python
arr1 = np.array([[1], [2], [3], [4]])
arr2 = np.array([[1, 2, 3, 4]])   # 2d array with single row
print(arr1.shape)
print(arr2.shape)
```

```python
arr3 = arr1 + arr2  # both have 1 so we can perfrom broadcasting but cant do if 1 is not their because 4 dont match
arr3
```

FILTERING

```python
ages = np.array([[21, 17, 19, 20, 19, 65],
                 [39, 22, 15, 99, 20, 21]])
```

```python
teen = ages[ages < 18]
adult = ages[(ages > 18) & (ages < 65) ] # and operator
print(teen)
print(adult)
```

```python
adult = np.where(ages >= 18, ages, 0) # it retain the exact shape of array , condition, array, value at the place were condition not true so 0
adult
```
