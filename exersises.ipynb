**Valid Sudoku**

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
# sum of each row
a = np.sum(s,axis= 0)
for rows in a :
  if rows != 45 :
    print("not valid sudoku: ")
    break
else :
  print("valid sudoku")
```

```python
# sum of each column
a = np.sum(s,axis= 1)
for cloumn in a :
  if cloumn != 45 :
    print("not valid sudoku: ")
    break
else :
  print("valid sudoku")
```

```python
# check the sum in 3*3 matrix should be equal to 45

for rows in range(0, 9, 3) :
  for col in range(0, 9, 3) :
    a = np.sum(s[rows: rows + 3, col : col + 3])
    if( a != 45) :
      print("sudoku not valid: ")
      break
else :
  print("sudoku valid")
```

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
data.shape
```

```python
#Find the average age of students.
np.mean(data[:,0])  # select every row and 0th cloumn , give exact value the one you want to select , in slicing the last value should be +1 
```

```python
#Extract Math marks of all students.
a = data[:, 1]
print(a)
```

```python
#Find the highest Science mark.
high = np.max(data[:,2])
print(high)
```

```python
#Get details of the student who scored more than 90 in Math.
index = 0
for i in data[:, 1] :
  if i > 90 :
    print(f"Student {index} score more than 90")
  index = index + 1

    
```

```python
#Increase Math marks of all students by 5.
for  index, marks in enumerate(data[:, 1]) :
  data[index, 1] = marks + 5 

print(data)
print(" ")

# another way

data[:,1] += 5
print(data)
 
```

```python
#Find how many students are younger than 19.
for student, age in enumerate(data[:, 0]) :
  if age < 19 :
    print(f"Student {student} is younger than 19")
```

```python
#Calculate the average marks in each subject (column-wise mean).
avg = np.mean(data[:, 1:], axis=1)
print(avg)
```
